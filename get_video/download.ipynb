```python
import subprocess

url = "https://www.youtube.com/watch?v=M6WSW4rjjsE"
output_path = "video.mp4"

# Download video using yt-dlp
subprocess.run([
    "yt-dlp",
    "-f", "bestvideo[ext=mp4]+bestaudio[ext=m4a]/best[ext=mp4]/best",
    "-o", output_path,
    url
])
```
