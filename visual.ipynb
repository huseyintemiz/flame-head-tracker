```python
output_img_org = Image.open('images/output_img_org.png')
output_aligned = Image.open('images/output_aligned.png')

# Get dimensions and calculate aspect ratios
img_org_size = output_img_org.size
img_aligned_size = output_aligned.size

# Calculate the target height (use the smaller height or set a fixed height)
target_height = min(img_org_size[1], img_aligned_size[1])

# Resize images to have the same height while maintaining aspect ratio
img_org_ratio = img_org_size[0] / img_org_size[1]
img_aligned_ratio = img_aligned_size[0] / img_aligned_size[1]

output_img_org_resized = output_img_org.resize((int(target_height * img_org_ratio), target_height), Image.Resampling.LANCZOS)
output_aligned_resized = output_aligned.resize((int(target_height * img_aligned_ratio), target_height), Image.Resampling.LANCZOS)

# fig, ax = plt.subplots(1, 2, figsize=(10, 5))
fig, ax = plt.subplots(1, 2, figsize=(15, 5))
# Adjust subplot spacing to accommodate different widths
plt.subplots_adjust(wspace=0.1)
ax[0].imshow()
ax[0].set_title('Input (Matted)')
ax[0].axis('off')
ax[1].imshow(output_aligned_resized)
ax[1].set_title('Aligned (Matted)')
ax[1].axis('off')
plt.show()
```
